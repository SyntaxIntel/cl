# Clonal Selection Algorithm (CSA)

This notebook implements a simple and explainable version of the Clonal Selection Algorithm (CSA) in Python.

CSA is inspired by the natural immune system and is used for optimization problems. It works by generating a population of candidate solutions (antibodies), selecting the best ones, cloning them, introducing mutations, and selecting the best individuals for the next generation.

```python
import numpy as np


# Objective function: Sphere function (minimize sum of squares)
def objective(x):
    return np.sum(x**2)
```

## Step 1: Initialize Population
Generate a random population of candidate solutions (antibodies).

```python
def initialize_population(pop_size, dim, lower, upper):
    return np.random.uniform(lower, upper, (pop_size, dim))


pop_size = 10
dim = 2
lower, upper = -5, 5
population = initialize_population(pop_size, dim, lower, upper)
population
```

## Step 2: Evaluate Fitness
Calculate the fitness (objective value) for each antibody.

```python
fitness = np.array([objective(ind) for ind in population])
fitness
```

## Step 3: Select Best Antibodies
Select the top N antibodies with the best (lowest) fitness.

```python
num_selected = 4
selected_indices = np.argsort(fitness)[:num_selected]
selected = population[selected_indices]
selected
```

## Step 4: Clone and Mutate
Clone the selected antibodies and introduce random mutations.

```python
def clone_and_mutate(selected, clones_per_antibody, mutation_rate=0.2):
    clones = []
    for antibody in selected:
        for _ in range(clones_per_antibody):
            clone = antibody.copy()
            mutation = np.random.normal(0, mutation_rate, size=antibody.shape)
            clone += mutation
            clones.append(clone)
    return np.array(clones)


clones_per_antibody = 3
clones = clone_and_mutate(selected, clones_per_antibody)
clones
```

## Step 5: Select Next Generation
Combine original population and clones, select the best individuals for the next generation.

```python
combined = np.vstack((population, clones))
combined_fitness = np.array([objective(ind) for ind in combined])
next_gen_indices = np.argsort(combined_fitness)[:pop_size]
population = combined[next_gen_indices]
population
```

## Step 6: Repeat
Repeat the process for a number of generations.

```python
generations = 10
for gen in range(generations):
    fitness = np.array([objective(ind) for ind in population])
    selected_indices = np.argsort(fitness)[:num_selected]
    selected = population[selected_indices]
    clones = clone_and_mutate(selected, clones_per_antibody)
    combined = np.vstack((population, clones))
    combined_fitness = np.array([objective(ind) for ind in combined])
    next_gen_indices = np.argsort(combined_fitness)[:pop_size]
    population = combined[next_gen_indices]
    print(
        f"Generation {gen + 1}, Best Fitness: {combined_fitness[next_gen_indices[0]]:.4f}"
    )
```
