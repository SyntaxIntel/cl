## 4. Set Genetic Algorithm Parameters
Define population size and number of generations.

```python
POPULATION_SIZE = 10
GENERATIONS = 5
```

## 5. Create Fitness and Individual Types
Set up the DEAP framework for minimization and individual representation.

```python
creator.create("FitnessMin", base.Fitness, weights=(-1.0,))
creator.create("Individual", list, fitness=creator.FitnessMin)
```

## 6. Initialize Toolbox and Register Functions
Define how individuals are created and how genetic operators work.

```python
toolbox = base.Toolbox()
# Attribute generators
toolbox.register("attr_neurons", random.randint, 1, 100)  # Number of neurons
toolbox.register("attr_layers", random.randint, 1, 5)  # Number of layers
# Structure initializers
# Each individual: [neurons, layers]
toolbox.register(
    "individual",
    tools.initCycle,
    creator.Individual,
    (toolbox.attr_neurons, toolbox.attr_layers),
    n=1,
)
toolbox.register("population", tools.initRepeat, list, toolbox.individual)
# Genetic operators
# Evaluation
# Crossover
# Mutation
# Selection
toolbox.register("evaluate", evaluate)
toolbox.register("mate", tools.cxTwoPoint)
toolbox.register("mutate", tools.mutUniformInt, low=1, up=100, indpb=0.2)
toolbox.register("select", tools.selTournament, tournsize=3)
```

## 7. Create Initial Population
Generate the initial set of individuals.

```python
population = toolbox.population(n=POPULATION_SIZE)
```

## 8. Run the Genetic Algorithm
Evolve the population over several generations.

```python
for gen in range(GENERATIONS):
    # Variation: crossover and mutation
    offspring = algorithms.varAnd(population, toolbox, cxpb=0.5, mutpb=0.1)
    # Evaluate fitness
    fitnesses = toolbox.map(toolbox.evaluate, offspring)
    for ind, fit in zip(offspring, fitnesses):
        ind.fitness.values = fit
    # Select the next generation
    population = toolbox.select(offspring, k=len(population))
```

## 9. Get and Display the Best Solution
Select and print the best individual (parameter set) found.

```python
best_individual = tools.selBest(population, k=1)[0]
best_params = best_individual
print("Best Parameters:", best_params)
```
