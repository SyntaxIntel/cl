# Structure Damage Classification using Artificial Immune Recognition System (AIRS)

This notebook demonstrates a simple, explainable implementation of the AIRS algorithm for classifying structural damage. The code is designed to be easy to understand and present to an instructor.

```python
import numpy as np


# Step 1: Generate dummy data (replace with real data for actual use)
def generate_dummy_data(samples=100, features=10):
    data = np.random.rand(samples, features)
    labels = np.random.randint(0, 2, size=samples)  # 0: No Damage, 1: Damage
    return data, labels


# Step 2: Define a simple AIRS class
class AIRS:
    def __init__(self, num_detectors=10, hypermutation_rate=0.1):
        self.num_detectors = num_detectors
        self.hypermutation_rate = hypermutation_rate
        self.detectors = None

    def train(self, X, y):
        # Randomly select detectors from training data
        indices = np.random.choice(len(X), self.num_detectors, replace=False)
        self.detectors = X[indices]

    def predict(self, X):
        predictions = []
        for sample in X:
            # Calculate distance from sample to each detector
            distances = np.linalg.norm(self.detectors - sample, axis=1)
            # Assign label based on nearest detector (for demo, just use index parity)
            nearest = np.argmin(distances)
            label = nearest % 2  # Simple mapping for demonstration
            predictions.append(label)
        return np.array(predictions)


# Step 3: Prepare data
data, labels = generate_dummy_data(samples=100, features=10)
split_ratio = 0.8
split_index = int(split_ratio * len(data))
train_data, test_data = data[:split_index], data[split_index:]
train_labels, test_labels = labels[:split_index], labels[split_index:]

# Step 4: Train AIRS
airs = AIRS(num_detectors=10, hypermutation_rate=0.1)
airs.train(train_data, train_labels)

# Step 5: Predict and evaluate
predictions = airs.predict(test_data)
accuracy = np.mean(predictions == test_labels)
print(f"Accuracy: {accuracy:.2f}")
```

## Explanation
- **Dummy Data**: Randomly generated for demonstration. Replace with real structural data for practical use.
- **AIRS Algorithm**: Mimics immune system pattern recognition. Here, detectors are randomly chosen samples.
- **Prediction**: Assigns class based on the closest detector. For simplicity, label is determined by detector index parity.
- **Accuracy**: Compares predicted labels to true labels.

This code is intentionally simple for exam and explanation purposes.
