```python
# 1. Data Extraction
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris

# Load the Iris dataset (Extraction)
iris = load_iris()
```

## 1. Data Transformation

```python
# Prepare DataFrame for easier plotting
iris_df = pd.DataFrame(iris.data, columns=iris.feature_names)
iris_df["species"] = pd.Categorical.from_codes(iris.target, iris.target_names)
```

## 2. Data Visualization

Let's explore the Iris dataset with several visualizations to understand its features and species distribution.

### Example 1: Sepal Length vs Sepal Width (Scatter Plot)
This plot shows the relationship between sepal length and sepal width for each Iris species.

```python
plt.figure(figsize=(7, 5))
for i, name in enumerate(iris.target_names):
    plt.scatter(
        iris.data[iris.target == i, 0], iris.data[iris.target == i, 1], label=name
    )
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Sepal Width (cm)")
plt.title("Sepal Length vs Sepal Width by Species")
plt.legend()
plt.show()
```

### Example 2: Petal Length Distribution (Box Plot)
This box plot visualizes the distribution of petal lengths for each species.

```python
plt.figure(figsize=(7, 4))
plt.boxplot(
    [iris.data[iris.target == i, 2] for i in range(3)],
    labels=iris.target_names,
    notch=True,
    patch_artist=True,
)
plt.xlabel("Species")
plt.ylabel("Petal Length (cm)")
plt.title("Distribution of Petal Length by Species")
plt.show()
```

### Example 3: Histogram of Sepal Width
This histogram shows the frequency distribution of sepal width for each species.

```python
plt.figure(figsize=(7, 4))
for i, name in enumerate(iris.target_names):
    plt.hist(iris.data[iris.target == i, 1], alpha=0.5, label=name)
plt.xlabel("Sepal Width (cm)")
plt.ylabel("Frequency")
plt.title("Histogram of Sepal Width by Species")
plt.legend()
plt.show()
```

### Example 4: Pairplot (Scatter Matrix)
A scatter matrix provides a quick overview of pairwise relationships between all features, colored by species.

```python
pd.plotting.scatter_matrix(
    iris_df, figsize=(10, 8), c=iris.target, cmap="plasma", diagonal="hist"
)
plt.suptitle("Pairwise Feature Relationships in Iris Dataset")
plt.show()
```
